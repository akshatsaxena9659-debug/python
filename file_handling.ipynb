```python
#opening and reading file
file = open('data.txt','r')
content = file.read()
print(content)
file.close()
```

```python
#opening and writing to file
file = open('data.txt','w')
file.write('How r u?')
file.close()
```

```python
#opening and adding text with existing content
file = open('data.txt','a')
file.write('\nNew record added')
file.close()
```

```python
#no need to close the file manually when using with statement
with open('data.txt','r') as file:
    content = file.read()
    print(content)
```

FILE PROPERTIES

```python
file = open('data.txt','r')
print(file.name)
print(file.mode)
print(file.closed)
print(file.tell()) #cursor position
file.read(5)
print(file.tell())
file.seek(8) #changes cursor position 
print(file.tell())
```

EXCEPTION HANDLING

```python
age=int(input("Enter your age:"))
if age>18:
    print("Valid")
else:
    print("invalid")
```

```python
try:
    age=int(input("Enter your age:"))
    print(age)
except ValueError:
    print("Please enter a valid age")
```

```python
try:
    file = open('newfile.txt','x')
    print("File created successfully.")
except FileExistsError:
    print("File already exists.")
finally:
    file.close()
```

```python
try:
    num = 10
    result = num/0
    print(result)
except ZeroDivisionError:
    print("You cannot divide by zero")
```
