```python
n= int(input("Enter a number: "))
count = 0
for i in range(1, n+1):
    if i%5 == 0:
        count += 1
print(count)
```

```python
#lambda function ka use krke 
n = int(input("Enter a number"))
check = lambda x:x%5==0
count=0
for i in range(1,n+1):
      if check(i):
          count+=1
print(count)
```

- take 3 digit number and print the last digit only 
- check string starts with capital letter or not 
- take a number and check it is a multiple of 3 or 7 
- check if a character is digit or not 
- calculate discount (10% if amount >5000)

```python
x=int(input("Enter a three digit number"))
check = lambda x:x%10
if check(x):
    print(check(x))
```

```python
x= str(input("Enter a three digit number"))
check = lambda x: x[0].isupper()
if check(x):
    print(check(x), end=" ")
else:
    print("not upper case")
```

```python
x=int(input("Enter a number"))
check = lambda x: x%3==0 or x%7==0
if check(x):
    print(check(x), end=" ")
else:
    print("not divisible by 3 or 7")
```

```python
x = input("Enter a number: ")
check = lambda x: x.isdigit()
if check(x):
    print(check(x))
else:   
    print("not a number")
```

```python
x= int(input("Enter the amount"))
check = lambda x: "Prime" if x>1 and all(x%i!=0 for i in range(2,x)) else "Not Prime"
print(check(x))
```
