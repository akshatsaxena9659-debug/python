```python
n= int(input("Enter a number: "))
count = 0
for i in range(1, n+1):
    if i%5 == 0:
        count += 1
print(count)
```

```python
#lambda function ka use krke 
n = int(input("Enter a number"))
check = lambda x:x%5==0
count=0
for i in range(1,n+1):
      if check(i):
          count+=1
print(count)
```
